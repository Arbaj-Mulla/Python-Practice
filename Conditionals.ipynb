## Testing truth value

### `bool`

```python
print(f"type of True and False: {type(True)}")
```

```python
print(f"0: {bool(0)}, 1: {bool(1)}")
print(f"empty list: {bool([])}, list with values: {bool(['woop'])}")
print(f"empty dict: {bool({})}, dict with values: {bool({'Python': 'cool'})}")
```
